# -- Purpose : Common Data Quality Validation Functions--

```python
#Duplicate check

from pyspark.sql.functions import *

def duplicate_check(df, column):

    return (

        df.groupBy(column)

        .count()

        .filter(col("count") > 1)

    )
```

```python
#Null Summary

def null_summary(df):

    return (

        df.select(

            [

                sum(

                    when(col(c).isNull(),1)

                    .otherwise(0)

                ).alias(c)

                for c in df.columns

            ]

        )

    )
```
