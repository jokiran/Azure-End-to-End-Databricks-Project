```python
print("===== SILVER LAYER VALIDATION =====")

silver_employee_df = spark.table(
    "databricks_project1.silver.employee_payroll"
)

silver_labor_df = spark.table(
    "databricks_project1.silver.labor_position"
)

silver_facility_df = spark.table(
    "databricks_project1.silver.facility"
)

print("Silver Employee count:", silver_employee_df.count())
print("Silver Labor Position count:", silver_labor_df.count())
print("Silver Facility count:", silver_facility_df.count())
```

Timestamps

```python
from pyspark.sql import functions as F
print(
    "Silver Employee max timestamp:",
    silver_employee_df
        .select(F.max("ingestion_timestamp"))
        .first()[0]
)

print(
    "Silver Labor Position max timestamp:",
    silver_labor_df
        .select(F.max("ingestion_timestamp"))
        .first()[0]
)

print(
    "Silver Facility max timestamp:",
    silver_facility_df
        .select(F.max("ingestion_timestamp"))
        .first()[0]
)
```

verify watermarks

```python
watermark_df = spark.table(
    "databricks_project1.silver.pipeline_watermark"
)

display(
    watermark_df
    .orderBy("pipeline_name", "source_name")
)
```

Verify Silver Employee deduplication

```python
print(
    "Duplicate Employee_Code rows:",
    silver_employee_df
        .groupBy("Employee_Code")
        .count()
        .filter(F.col("count") > 1)
        .count()
)
```

Verify Silver Labor Position deduplication

```python
print(
    "Duplicate Labor_Position_Code rows:",
    silver_labor_df
        .groupBy("Labor_Position_Code")
        .count()
        .filter(F.col("count") > 1)
        .count()
)
```
